```python
from tensorflow.keras.models import load_model
from os import path
from os import listdir
import numpy as np
import tensorflow as tf
import requests
from PIL import Image
from io import BytesIO
import pickle

model = load_model('pokemon_model.keras')

classes = []
for _Class in listdir('/Users/remboymitch/Downloads/pokemon'): #replace with personal device path
    classes.append(_Class)
classes.sort()

with open('pokemon_classes', 'wb') as pickle_out:
    pickle.dump(classes, pickle_out)

print(classes)

def predict(pokemon_image):
    image = pokemon_image.resize((128, 128))
    img_array = np.array(image).astype('float32')
    img_array = tf.keras.applications.mobilenet_v2.preprocess_input(img_array)
    img_array = np.expand_dims(img_array, axis=0)
    prediction = model.predict(img_array)

    pred = model.predict(img_array)
    print(pred)
    print(np.argmax(pred))

    return np.argmax(prediction, axis = 1).tolist()[0]

def identify(img_url):
    page = requests.get(img_url)
    poke_img = Image.open(BytesIO(page.content)).convert('RGB')
    poke_img = poke_img.resize((128, 128))
    classes = pickle.load(open('pokemon_classes', 'rb'))
    print("A wild {} appears!".format(classes[predict(poke_img) + 1]))
    print("\n\n\n")
    poke_img.show()
```

```python
url = input("Please enter the url for pokemon image: ")
print()
identify(url)
```
